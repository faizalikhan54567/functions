##**1.What is the difference between a function and a method in Python?**

  -In Python, the **main difference between a function and a method** is:

###  Function
- A **function** is a block of reusable code that performs a specific task.
- It is defined using the `def` keyword or as a lambda expression.
- It can exist **independently** of any object.
- Example:
  ```python
  def greet(name):
      return f"Hello, {name}!"
  ```

###  Method
- A **method** is a function that is **associated with an object**—typically defined inside a class.
- It automatically takes the object instance (`self`) as its first argument (for instance methods).
- Methods are **called on objects**.
- Example:
  ```python
  class Greeter:
      def greet(self, name):
          return f"Hello, {name}!"
          
  g = Greeter()
  print(g.greet("Alice"))  # greet is a method here
  ```

##**2.Explain the concept of function arguments and parameters in Python?**

  -In Python, **parameters** and **arguments** are closely related but refer to different things:

---

###  Parameters
- **Parameters** are **variables listed in the function definition**.
- They act as **placeholders** for the values the function will receive when it is called.
- Example:
  ```python
  def greet(name):  # 'name' is a parameter
      print(f"Hello, {name}!")
  ```

---

###  Arguments
- **Arguments** are the **actual values** you pass to the function when you call it.
- Example:
  ```python
  greet("Alice")  # "Alice" is an argument
  ```

---

###  Types of Function Parameters (and corresponding arguments)
Python supports several kinds of parameters/arguments:

| Type                 | Syntax                          | Example                                 |
|----------------------|----------------------------------|------------------------------------------|
| **Positional**        | Based on position               | `def add(x, y):` → `add(2, 3)`           |
| **Keyword**           | Named explicitly                | `add(x=2, y=3)`                          |
| **Default**           | With default values             | `def greet(name="Guest"):`              |
| **Variable-length**   | Accepts many args               | `*args`, `**kwargs`                      |

Example using all:
```python
def example(a, b=2, *args, **kwargs):
    print(a, b, args, kwargs)

example(1, 3, 4, 5, key="value")
# Output: 1 3 (4, 5) {'key': 'value'}
```

##**4.What is the purpose of the `return` statement in a Python function?**

-The `return` statement in a Python function is used to **send a result back to the caller** of the function. It's how a function **outputs a value**.

---

###  **Key Purposes of `return`:**

i. **Returns a value to the caller**:
   ```python
   def add(a, b):
       return a + b

   result = add(3, 5)  # result = 8
   ```

ii. **Ends the function’s execution** immediately:
   - Code after `return` won’t run.
   ```python
   def check():
       return "Done"
       print("This won't run")
   ```

iii. **Can return multiple values** as a tuple:
   ```python
   def stats(x, y):
       return x + y, x * y

   sum_, product = stats(2, 3)  # sum_ = 5, product = 6
   ```

iv. **Return nothing (implicitly returns `None`)**:
   ```python
   def say_hi():
       print("Hi!")

   result = say_hi()  # result is None
   ```

##**5. What are iterators in Python and how do they differ from iterables?**

 -In Python, **iterators** and **iterables** are closely related but serve different roles in iteration.

---

##  **Iterable**
- An **iterable** is any Python object capable of returning its members one at a time.
- Examples: `list`, `tuple`, `str`, `set`, `dict`, etc.
- It implements the `__iter__()` method.
- Can be **passed to `iter()`** to get an iterator.

```python
my_list = [1, 2, 3]      # This is an iterable
iterator = iter(my_list) # Getting an iterator from the iterable
```

---

##  **Iterator**
- An **iterator** is an object that **represents a stream of data**.
- It implements both:
  - `__iter__()` (returns itself)
  - `__next__()` (returns the next item and raises `StopIteration` when exhausted)

```python
iterator = iter([10, 20, 30])
print(next(iterator))  # 10
print(next(iterator))  # 20
```

---

##  **Key Differences**

| Feature       | Iterable                         | Iterator                            |
|---------------|----------------------------------|-------------------------------------|
| Purpose       | Can be looped over               | Generates values one at a time      |
| Methods       | Must implement `__iter__()`      | Implements `__iter__()` and `__next__()` |
| Memory        | Often stores all elements        | Can be more memory-efficient        |
| Usage         | Used in loops (e.g., `for`)      | Used with `next()` manually or in loops |
| Examples      | `list`, `dict`, `str`, `set`     | Object returned by `iter()` or generator |

##**6. Explain the concept of generators in Python and how they are defined?**

  -In Python, **generators** are a simple and memory-efficient way to create **iterators**. They allow you to iterate over a sequence of values **lazily**, meaning values are produced **one at a time** and **only when needed**, rather than storing the entire sequence in memory.

---


A **generator** is:
- A type of **iterator**.
- Defined using either:
  - A **function with `yield`**, or
  - A **generator expression**.

---

##  i. **Generator Function (`yield`)**
   -A function that **uses `yield` instead of `return`**.
- Each time `yield` is called, the function **pauses** and resumes later where it left off.

###  Example:
```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1

gen = countdown(3)
print(next(gen))  # 3
print(next(gen))  # 2
```

---

##  ii. **Generator Expression**
- Like a list comprehension, but uses **parentheses** `()` instead of square brackets `[]`.
- More memory-efficient.

### Example:
```python
squares = (x * x for x in range(5))
print(next(squares))  # 0
print(next(squares))  # 1
```

---

##**7.What are the advantages of using generators over regular functions?**


  -Generators in Python offer several advantages over regular functions, especially when working with large datasets or streams of data. Here’s a breakdown of their key benefits:

---

## **Advantages of Generators**

### i. **Memory Efficiency**
- **Generators don’t store all values in memory**—they yield one value at a time.
- Great for processing large or infinite sequences.
  
```python
def big_range():
    for i in range(1_000_000):
        yield i  # uses far less memory than returning a list
```

---

### ii. **Lazy Evaluation**
- Values are **computed only when needed**, which can speed up startup time and reduce computation when only part of the data is required.

```python
gen = (x * x for x in range(1000000))
print(next(gen))  # computes only the first value
```

---

### iii. **State Preservation Between Iterations**
- Each `yield` **suspends execution**, preserving the function’s local state between calls.

```python
def counter():
    x = 0
    while True:
        yield x
        x += 1  # remembers x across calls
```

---

### iv. **Cleaner Code for Iteration**
- Makes code simpler and easier to read when building iterators manually (no need for `__iter__` and `__next__`).

---

### v. **Pipeline Composition**
- Generators can be **chained** together to build data processing pipelines, similar to Unix pipes.

```python
def read_lines(file):
    for line in file:
        yield line.strip()

def filter_empty(lines):
    for line in lines:
        if line:
            yield line
```

##**8.What is a lambda function in Python and when is it typically used?**

 -A **lambda function** in Python is a **small, anonymous function** defined using the `lambda` keyword instead of `def`. It is often used for **short, simple operations** where defining a full function would be unnecessary or verbose.

---

 **Syntax of Lambda Function**

```python
lambda arguments: expression
```

### Example:
```python
square = lambda x: x * x
print(square(5))  # Output: 25
```

---

 **When to Use Lambda Functions**

Lambda functions are typically used when:
i. **A short function is needed for a short period of time**.
ii. **Used as an argument to higher-order functions** like `map()`, `filter()`, and `sorted()`.

---

###  Common Use Cases:

#### i. **With `map()`**
```python
nums = [1, 2, 3, 4]
squares = list(map(lambda x: x * x, nums))
```

#### ii. **With `filter()`**
```python
even = list(filter(lambda x: x % 2 == 0, nums))
```

#### iii. **With `sorted()`**
```python
data = [('apple', 2), ('banana', 1)]
sorted_data = sorted(data, key=lambda x: x[1])
```

#### iv. **In GUI Callbacks or Event Handlers**
```python
button.config(command=lambda: print("Clicked!"))
```

#**9.Explain the purpose and usage of the `map()` function in Python?**

   The `map()` function in Python is used to **apply a function to every item** in an iterable (like a list or tuple) and return a **map object** (which is an iterator) with the results.

---

##  **Purpose of `map()`**
To **transform** elements of an iterable by applying a function to each element **without writing a loop manually**.

-  **Syntax**
```python
map(function, iterable)
```

- `function`: A function that takes one argument.
- `iterable`: A sequence (like a list, tuple, etc.).

---  **Example: Using `map()` with a Lambda**

```python
nums = [1, 2, 3, 4]
squares = map(lambda x: x * x, nums)
print(list(squares))  # [1, 4, 9, 16]
```

---

  **Example: Using `map()` with a Regular Function**

```python
def upper_case(s):
    return s.upper()

names = ['alice', 'bob']
result = map(upper_case, names)
print(list(result))  # ['ALICE', 'BOB']
```

---
**Multiple Iterables**

You can also use `map()` with **multiple iterables**, as long as the function accepts that many arguments:

```python
a = [1, 2, 3]
b = [10, 20, 30]
result = map(lambda x, y: x + y, a, b)
print(list(result))  # [11, 22, 33]
```

##**10.What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?**

  -Great question! The functions `map()`, `reduce()`, and `filter()` in Python are part of a functional programming toolkit. They each serve different purposes for processing iterables. Here's a clear comparison:

---

## . `map(function, iterable)`
### **Purpose:** Transform each item in an iterable.

- **Applies a function to every element** of the iterable.
- Returns a `map` object (iterator) of the results.

###  Example:
```python
nums = [1, 2, 3]
squares = map(lambda x: x * x, nums)
print(list(squares))  # [1, 4, 9]
```

---

##  ii. `filter(function, iterable)`
### **Purpose:** Filter items based on a condition.

- **Keeps elements** for which the function returns `True`.
- Returns a `filter` object (iterator) of the filtered items.

###  Example:
```python
nums = [1, 2, 3, 4]
evens = filter(lambda x: x % 2 == 0, nums)
print(list(evens))  # [2, 4]
```

---

##  iii. `reduce(function, iterable)`
### **Purpose:** Reduce a sequence to a **single value** by applying a function cumulatively.

- Must be **imported** from `functools`.
- Applies the function cumulatively to the items, from left to right.

### Example:
```python
from functools import reduce

nums = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, nums)
print(product)  # 24
```

---

| Function  | Purpose                  | Returns       | Example Use              |
|-----------|--------------------------|---------------|---------------------------|
| `map()`   | Transform items           | map object    | Square each number        |
| `filter()`| Filter items by condition | filter object | Keep only even numbers    |
| `reduce()`| Aggregate to single value | single value  | Compute sum or product    |
